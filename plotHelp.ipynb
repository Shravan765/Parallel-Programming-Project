```python
import matplotlib.pyplot as plt
```

```python
plt.figure()
plt.title("Graph 1: Varying Number of Threads (Size of array = 10000000)")
plt.xlabel("Number of threads")
plt.ylabel("Time in milliseconds")
xThreads = [2,4,8,16,32]
ymerge = [1303, 920, 535, 460, 452]
yQuick = [1027, 900, 648, 592, 554]
yOddEven = [1429, 1136, 862, 767, 1280]
plt.plot(xThreads, ymerge, marker='*',color="#6200FF" ,label="Parallel MergeSOrt")
plt.plot(xThreads, yQuick, marker='*',color="#1A8080" ,label="Parallel QuickSort")
plt.plot(xThreads, yOddEven, marker='*',color="#AB912B" ,label="Parallel OddEvenSort")
plt.legend()
plt.show()
plt.savefig("1_thread.jpg")
```

The exceptionally poor performance of odd even at 32 threads is due to thread creation destruction overhead surpassing the gains due to parallelism

```python
plt.figure()
plt.title("Graph 2: Varying Size of input (Number of threads=16)")
plt.xlabel("Size of inputs")
plt.ylabel("Time in milliseconds")
xThreads = [1000000, 5000000, 10000000, 50000000, 100000000]
ymerge = [72, 216, 442, 2526, 5101]
yQuick = [130, 260, 539, 3389, 7771]
yOddEven = [146, 349, 769, 4127, 10579]
plt.plot(xThreads, ymerge, marker='*',color="#6200FF" ,label="Parallel MergeSOrt")
plt.plot(xThreads, yQuick, marker='*',color="#1A8080" ,label="Parallel QuickSort")
plt.plot(xThreads, yOddEven, marker='*',color="#AB912B" ,label="Parallel OddEvenSort")
plt.legend()
plt.show()
plt.savefig("2_thread.jpg")
```

Massive increase from 10M to 50M is due to cache limits, specifically,  
10M = 40MB  
50M = 200 MB  
  
So data is not directly coming from cache
